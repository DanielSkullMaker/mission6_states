```python
from PIL import Image
import os
```

```python
#Считаем число файлов в каждой папке

path, dirs, files = next(os.walk("/Users/vladkorsikov/research/dataset_main/glioma_raw"))
glioma_count = len(files)
print(glioma_count)

path, dirs, files = next(os.walk("/Users/vladkorsikov/research/dataset_main/meningioma_raw"))
meningioma_count = len(files)
print(meningioma_count)

path, dirs, files = next(os.walk("/Users/vladkorsikov/research/dataset_main/pituitary_raw"))
pituitary_count = len(files)
print(pituitary_count)

path, dirs, files = next(os.walk("/Users/vladkorsikov/research/dataset_main/test_raw"))
test_count = len(files) -1
print(test_count)
```

```python
#Задаем функцию


def resize_image_with_format(image_path, new_width, new_height, output_path):
    try:
        img = Image.open(image_path)
        img = img.resize((new_width, new_height), Image.LANCZOS)
        img.save(output_path, format=img.format)  # Сохраняет в том же формате, что и исходник
        print(f"Изображение успешно изменено и сохранено как {output_path}")
    except FileNotFoundError:
        print(f"Файл {image_path} не найден.")
    except Exception as e:
        print(f"Произошла ошибка: {e}")

```

```python
#задаем ширину и высоту (в пикселях)
new_width = 256 
new_height = 256

glioma_image_numbers = []
glioma_save_image_numbers = []
meningioma_image_numbers = []
meningioma_save_image_numbers = []
pituitary_image_numbers = []
pituitary_save_image_numbers = []
test_image_numbers = []
test_save_image_numbers = []

#Местонахождение папок
path_glioma = '/Users/vladkorsikov/research/dataset_main/glioma_raw'
path_glioma_save = '/Users/vladkorsikov/research/dataset_main/2_glioma_resize'
path_meningioma = '/Users/vladkorsikov/research/dataset_main/meningioma_raw'
path_meningioma_save = '/Users/vladkorsikov/research/dataset_main/2_meningioma_resize'
path_pituitary = '/Users/vladkorsikov/research/dataset_main/pituitary_raw'
path_pituitary_save = '/Users/vladkorsikov/research/dataset_main/2_pituitary_resize'
path_test = '/Users/vladkorsikov/research/dataset_main/test_raw'
path_test_save = '/Users/vladkorsikov/research/dataset_main/2_test_resize'

#заполняем списки имен
count = 0
for i in range(glioma_count):
  count += 1
  image_name = f'{path_glioma}/glioma{str(count)}.jpg'
  glioma_image_numbers.append(image_name)
  image_save_name = f'{path_glioma_save}/glioma{str(count)}.png'
  glioma_save_image_numbers.append(image_save_name)


count = 0
for i in range(meningioma_count):
  count += 1
  image_name = f'{path_meningioma}/meningioma{str(count)}.jpg'
  meningioma_image_numbers.append(image_name)
  image_save_name = f'{path_meningioma_save}/meningioma{str(count)}.png'
  meningioma_save_image_numbers.append(image_save_name)

count = 0
for i in range(pituitary_count):
  count += 1
  image_name = f'{path_pituitary}/pituitary{str(count)}.jpg'
  pituitary_image_numbers.append(image_name)
  image_save_name = f'{path_pituitary_save}/pituitary{str(count)}.png'
  pituitary_save_image_numbers.append(image_save_name)

count = 0
for i in range(test_count):
  count += 1
  image_name = f'{path_test}/test{str(count)}.png'
  test_image_numbers.append(image_name)
  image_save_name = f'{path_test_save}/test{str(count)}.png'
  test_save_image_numbers.append(image_save_name)

print(glioma_save_image_numbers[-1])
print(meningioma_save_image_numbers[-1])
print(meningioma_image_numbers[-1])
print(pituitary_save_image_numbers[-1])
print(test_save_image_numbers[-1])
```

```python
for image_path, output_path in zip(glioma_image_numbers, glioma_save_image_numbers):
    resize_image_with_format(image_path, new_width, new_height, output_path)

for image_path_2, output_path_2 in zip(meningioma_image_numbers, meningioma_save_image_numbers):
    resize_image_with_format(image_path_2, new_width, new_height, output_path_2)

for image_path_3, output_path_3 in zip(pituitary_image_numbers, pituitary_save_image_numbers):
    resize_image_with_format(image_path_3, new_width, new_height, output_path_3)

for image_path_4, output_path_4 in zip(test_image_numbers, test_save_image_numbers):
    resize_image_with_format(image_path_4, new_width, new_height, output_path_4)
```
