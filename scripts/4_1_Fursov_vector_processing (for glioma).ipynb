```python
with open('/Users/vladkorsikov/research/dataset_main/5_all_vectors/glioma/glioma_horizontal_vector.csv', newline='') as file:
    glioma_hor_vectors_0 = [row for row in csv.reader(file)]
```

```python
glioma_hor_vectors = []

for vector in glioma_hor_vectors_0:
    vector_add = []
    for bit in vector:
        bat = int(bit)
        vector_add.append(bat)
    glioma_hor_vectors.append(vector_add)

#Проверяем на случайном изображении
first_reference_number = 8
print(f'Первое эталонное изображение: glioma{first_reference_number+1}.png с номером вектора: {first_reference_number}')
```

```python
#зададим функцию поиска скалярного произведения
first = glioma_hor_vectors[18]
second = glioma_hor_vectors[19]
#print(first)
#print(hope_2)

#def optimization_vector_for_scalar_product (vector):
#    not_null_list = []
#    for number in vector:
#        if number > 0:
#            not_null_list.append(number)
#    return not_null_list



#optim_1 = optimization_vector_for_scalar_product(first)
#optim_2 = optimization_vector_for_scalar_product(second)

#print(len(optim_1))
#print(len(optim_2))


def scalar_product_vector_column_and_row(target_vector, list_vector):
    scal_product = 0
    for num_v, num_l in zip(target_vector, list_vector):
        num_v = int(num_v)
        num_l = int(num_l)
        increment=num_v*num_l
        scal_product +=increment
    return scal_product
        

scalar = scalar_product_vector_column_and_row(first, second)

print(scalar)

# теперь ищем норму выражения
def vector_norm(vector):
    norm_sq = 0
    for element in vector:
        element_sq = element**2
        norm_sq += element_sq
    norm = math.sqrt(norm_sq)
    norm = int(norm)
    return norm

norm_vec_1 = vector_norm(first)
norm_vec_2 = vector_norm(second)

def cos_criterea(scalar_product, norm_vec_1, norm_vec_2):
    norm_vec = norm_vec_1*norm_vec_2
    criterea = scalar_product/norm_vec
    return criterea



print(norm_vec_1)
print(norm_vec_2)
print(norm_vec_1*norm_vec_2)

final = cos_criterea(scalar, norm_vec_1, norm_vec_2)
print(final)

#scalar = scalar_product(example_image, example_image_2)
#print(scalar)
```

```python
trio_list = []
for analizing_number in range(len(glioma_hor_vectors)):
    first_reference = glioma_hor_vectors[analizing_number]
    first_reference_norm = vector_norm(first_reference)
    cos_criterea_list=[]
    for vector in glioma_hor_vectors:
        scalar_prod = scalar_product_vector_column_and_row(first_reference, vector)
        vector_normal = vector_norm(first_reference)
        criterea = cos_criterea(scalar_prod, first_reference_norm, vector_normal)
        cos_criterea_list.append(criterea)
    best_criteria = 999999999999
    criteria_number = -1
    current_best_number = 0
    for criteria in cos_criterea_list:
        criteria_number +=1
        if criteria < best_criteria:
            best_criteria = criteria
            current_best_number = criteria_number
    trio = [analizing_number, current_best_number, best_criteria]
    trio_list.append(trio)

print(trio_list)    
```

```python
trio_best_criteria = 999999999999
trio_criteria_number = -1
trio_current_best_number = []

for tr in trio_list:
    trio_criteria_number +=1
    if tr[2]<trio_best_criteria:
        trio_best_criteria = tr[2]
        trio_current_best_number = tr

print(trio_current_best_number)
```
