```python
#Необходимо найти для центров нового класса предельно отстоящий от них вектор, чтобы сформировать новые пары значений 
#Работает для 2 и более подклассов

import os
import numpy as np
import math
from time import time
from PIL import Image
import cv2
from matplotlib import pyplot as plt
import csv
import random
```

```python
#Импортируем вектора
with open('/Users/vladkorsikov/research/dataset_main/5_all_vectors/meningioma/meningioma_horizontal_vector.csv', newline='') as file:
    glioma_hor_vectors_0 = [row for row in csv.reader(file)]
glioma_hor_vectors_1 = []
count_v = 0

for vector in glioma_hor_vectors_0:
    vector_add = []
    for bit in vector:
        bat = int(bit)
        vector_add.append(bat)
    glioma_hor_vectors_1.append([count_v, vector_add])
    count_v +=1

print(len(glioma_hor_vectors_1))
```

```python
#Импортируем значение векторов
with open('/Users/vladkorsikov/research/dataset_main/5_all_vectors/meningioma/meningioma_class_core_vectors.csv', newline='') as file:
    glioma_core_vectors_0 = [row for row in csv.reader(file)]
glioma_core_vectors = []

for a in glioma_core_vectors_0:
    b = int(a[1])
    glioma_core_vectors.append(b)

print(glioma_core_vectors)
```

```python
#Записываем значения векторов, относящихся к центрам подклассов в отдельную переменную и убираем эти вектора из основного множества
glioma_hor_vectors_centers_0 = []
for core in glioma_core_vectors:
    count_l=0
    for vector in glioma_hor_vectors_1:
        if core == vector[0]:
            b = glioma_hor_vectors_1.pop(count_l)
            glioma_hor_vectors_centers_0.append(b)
        count_l+=1        

glioma_hor_vectors_centers=[]
for i in glioma_hor_vectors_centers_0:
    glioma_hor_vectors_centers.append(i[1])

resulting_element_list = []
glioma_hor_vectors = []


for i in glioma_hor_vectors_1:
    resulting_element_list.append(i[0])
    glioma_hor_vectors.append(i[1])
```

```python
#зададим функцию поиска скалярного произведения
first = glioma_hor_vectors[18]
second = glioma_hor_vectors[19]

def scalar_product_vector_column_and_row(target_vector, list_vector):
    scal_product = 0
    for num_v, num_l in zip(target_vector, list_vector):
        num_v = int(num_v)
        num_l = int(num_l)
        increment=num_v*num_l
        scal_product +=increment
    return scal_product
        

scalar = scalar_product_vector_column_and_row(first, second)

print(scalar)

# теперь ищем норму выражения
def vector_norm(vector):
    norm_sq = 0
    for element in vector:
        element_sq = element**2
        norm_sq += element_sq
    norm = math.sqrt(norm_sq)
    norm = int(norm)
    return norm

norm_vec_1 = vector_norm(first)
norm_vec_2 = vector_norm(second)

def cos_criterea(scalar_product, norm_vec_1, norm_vec_2):
    norm_vec = norm_vec_1*norm_vec_2
    criterea = scalar_product/norm_vec
    return criterea



print(norm_vec_1)
print(norm_vec_2)
print(norm_vec_1*norm_vec_2)

final = cos_criterea(scalar, norm_vec_1, norm_vec_2)
print(final)
```

```python
list_for_cos_criterea_list = []
max_number_list = []
for first_reference in glioma_hor_vectors_centers:
    first_reference_norm = vector_norm(first_reference)
    cos_criterea_list=[]
    count_master = 0
    for vector in glioma_hor_vectors:
       # optim_vector = optimization_vector_for_scalar_product(vector)
        scalar_prod = scalar_product_vector_column_and_row(first_reference, vector)
        vector_normal = vector_norm(first_reference)
        criterea = cos_criterea(scalar_prod, first_reference_norm, vector_normal)
        count_master=+1
        cos_criterea_list.append(criterea)
    max_number = max(cos_criterea_list)
    max_number_list.append(max_number)
    list_for_cos_criterea_list.append(cos_criterea_list)
    
    
print(len(list_for_cos_criterea_list))     
print(cos_criterea_list)  
print(max_number_list)
```

```python
#Определим очередность поиска пар в зависимости от значения максимально вектора

dct = {}
for i in range(len(glioma_core_vectors)):
	key = max_number_list[i]
	value = glioma_core_vectors[i]
	dct[key] = value
print(dct)

max_number_list.sort(reverse=True)
print(max_number_list)

glioma_core_vectors_2=[]
for i in max_number_list:
    glioma_core_vectors_2.append(dct[i])

print(glioma_core_vectors_2)

dict_for_cos_criterea_list = dict(zip(glioma_core_vectors, list_for_cos_criterea_list))

list_for_cos_criterea_list_2 = []
for i in glioma_core_vectors_2:
    list_for_cos_criterea_list_2.append(dict_for_cos_criterea_list[i])



#sorted_dct = sorted(dct.items(), key=lambda item: item[1], reverse = True)



#print(dct.type())
#keys = 
#keys = sorted_dct.keys()

#print(keys_list)

```
