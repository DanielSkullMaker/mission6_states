```python
# Импортируем функции из скрипта 5
# 1. Умножение транспонированной матрицы на обычный вектор-столбец
def tran_matrix_column_vector_product(reference_matrix, vector):
    product_tmcv = []
    product_matrix = []
    for reference_vector in reference_matrix:
        for number, number_2 in zip(vector, reference_vector):
            number_3 = number*number_2
            product_tmcv.append(number_3)
        product_matrix.append(product_tmcv)
        product_tmcv=[]

        # преобразуем шаг 1 в сумму элементов вектора
    first_new = []
    for vector in product_matrix:
        part = 0
        for number in vector:
            part +=number
        first_new.append(part)
    return first_new

#2. Транспонировали матрицу, полученную в предыдущем действии (производить преобразование программно не требуется)
# Значение матрицы взять из первого шага


# 3. Транспонируем эталонную матрицу и умножим ее на нетранспонированную версию
#умножение транспонированной матрицы на исходную подразумевает скалярное произведение каждого вектора на себя и остальные
#результаты записываются в строку
def tran_matrix_and_matrix_product(matrix):
    matrix_product_raw = []
    matrix_product_cell = []
    matrix_product = []
    count = 0
    for vector_1 in reference_matrix:
        for vector_2 in reference_matrix:
            product_cell = 0
            for a, b in zip(vector_1, vector_2):
                c = a*b
                product_cell += c
            matrix_product_raw.append(product_cell)
    for product in matrix_product_raw:
        if count < len(reference_matrix):
            matrix_product_cell.append(product)
            count+=1
        else:
            matrix_product.append(matrix_product_cell)
            matrix_product_cell = []
            matrix_product_cell.append(product)
            count=1
    matrix_product.append(matrix_product_cell)
    return matrix_product
        
#Ниже получаем значение третьего шага для каждого класса, которое будем использовать в цикле
#third_step = tran_matrix_and_matrix_product(reference_matrix) # Полученное значение мы будем использовать в итоговом цикле



# 4. Получаем обратную матрицу(значение постоянно для каждого класса, получаем список один раз и используем в цикле)

#fourth_step = np.linalg.inv(third_step) # Полученное значение мы будем использовать в итоговом цикле
#print(fourth_step)


# 5. Переносим функцию из предыдущего шага и находим скалярное произведение анализируемого вектора как строки на столбец
def scalar_product_vector_column_and_row(target_vector, list_vector):
    scal_product = 0
    for num_v, num_l in zip(target_vector, list_vector):
        num_v = int(num_v)
        num_l = int(num_l)
        increment=num_v*num_l
        scal_product +=increment
    return scal_product


#6. Находим итоговое значение

def final_step(first_step, fourth_step, fith_step):
    #умножаем вектор-строку на матрицу 2x2, для этого умножаем вектор на каждый столбец полученной матрицы
    sixth_one = []
    for vector in fourth_step:
        increm = 0
        for number_1, number_2 in zip(first_step, vector):
            number_3 = number_1 * number_2
            increm +=number_3
        sixth_one.append(increm)
    # Находим скалярное произведение полученного вектора-строки на вектор-столбец 
    sixth_two = 0
    for number_1, number_2 in zip(sixth_one, first_step):
        number_3 = number_1 * number_2
        sixth_two +=number_3
    #итоговое деление
    sixth_three = sixth_two/fith_step
    return sixth_three
```

```python
#Распределим вектора по классам в зависимости от количества элементов, которые нерёобходимо добавить
for idea in range(count_num):
    # Теперь найдем критерий сопряженности каждого вектора с каждым классом
    subclass_final_result = []
    for reference_matrix in glioma_subclass_vectors_2:
        third_step = tran_matrix_and_matrix_product(reference_matrix) # Полученное значение мы будем использовать во всем цикле с данным классом
        fourth_step = np.linalg.inv(third_step) # Полученное значение мы будем использовать в итоговом цикле
        final_results = []
        for vector_analise in glioma_hor_vectors:
            first_step = tran_matrix_column_vector_product(reference_matrix, vector_analise)
            fith_step = scalar_product_vector_column_and_row(vector_analise, vector_analise)
            final_new = final_step(first_step, fourth_step, fith_step)
            final_results.append(final_new)
        subclass_final_result.append(final_results)
    # Преобразуем в список, выведем сопряженность отдельных векторов к каждому классу в отдельные списки
    len_17 = len(subclass_final_result[0])
    subclass_final_result_2 = []
    for a in range(len_17):
        subclass_result = []
        for subclass in subclass_final_result:
            subclass_result.append(subclass[a])
        subclass_final_result_2.append(subclass_result)
    # Теперь проанализируем, как показатель сопряженности вектора для каждого подкласса соотносится со средним для данного класса значением
    #для этого создадим список средних начений
    subclass_final_result_middle = []
    for sub in subclass_final_result_2:
        count = 0
        b = 0
        for a in sub:
            b +=a
            count +=1
        c = b/count
        subclass_final_result_middle.append(c)
    #Найдем для каждого вектора соотношение показателя сопряженности для каждого подкласса к среднему значению
    subclass_final_result_3 = []
    count_vector = 0
    for sub in subclass_final_result_2:
        vector_sub = []
        for a in sub:
            b = a/subclass_final_result_middle[count_vector]
            vector_sub.append(b)
        subclass_final_result_3.append(vector_sub)
    # отберем для каждого подкласса по одному вектору с наибольшим соотношением сопряженности по отношению к другим векторам 
    best_of_list = []
    for i in range(len(glioma_subclass_vectors_2)):
        best_criteria = 0
        criteria_number = 0
        current_best_number_0 = -1
        for subclass in  subclass_final_result_3:
            criteria = subclass[i]
            if criteria > best_criteria:
                if criteria_number not in best_of_list:
                    best_criteria = criteria
                    current_best_number_0 = criteria_number
            criteria_number +=1
        best_of_list.append(current_best_number_0)
    # Сопоставим значение с номерами оставшихся векторов из списка номеров оставшихся векторов
    best_of_list_2 = []
    best_of_list_vector = []
    for i in best_of_list:
        vector_number = resulting_element_list[i]
        vector = glioma_hor_vectors[i]
        best_of_list_2.append(vector_number)
        best_of_list_vector.append(vector)
    print(best_of_list_2)   
    print(len(best_of_list_vector))
    print(len(best_of_list_vector[0]))
    #Удаляем взятые элементы из списков, из которых мы их взяли
    best_of_list.sort(key=None, reverse=True)
    print(best_of_list)
    for i in best_of_list:
        vector_number = resulting_element_list.pop(i)
        vector = glioma_hor_vectors.pop(i)
    print(resulting_element_list)
    print(len(resulting_element_list))
    #Вставим новые числа в подклассы
    #номера
    glioma_subclass_vectors_77 = []
    count = 0
    for subclass_number in glioma_subclass_vectors:
        subclass_number.append(best_of_list_2[count])
        glioma_subclass_vectors_77.append(subclass_number)
        count += 1
    glioma_subclass_vectors = glioma_subclass_vectors_77.copy()
    print(glioma_subclass_vectors)
    #значения векторов
    glioma_subclass_vectors_777 = []
    count = 0
    for vector_number in glioma_subclass_vectors_2:
        vector_number.append(best_of_list_vector[count])
        glioma_subclass_vectors_777.append(vector_number)
        count += 1
    glioma_subclass_vectors_2 = glioma_subclass_vectors_777.copy()
    gsv_2 = glioma_subclass_vectors_2[0]
    print(len(glioma_subclass_vectors_2))
    print(len(glioma_subclass_vectors_2[0]))
    print(len(gsv_2[0]))
```
