```python
# 1. Умножение транспонированной матрицы на обычный вектор-столбец

example_vector = glioma_hor_vectors[6]

def tran_matrix_column_vector_product(reference_matrix, vector):
    product_tmcv = []
    product_matrix = []
    for reference_vector in reference_matrix:
        for number, number_2 in zip(vector, reference_vector):
            number_3 = number*number_2
            product_tmcv.append(number_3)
        product_matrix.append(product_tmcv)
        product_tmcv=[]

        # преобразуем шаг 1 в сумму элементов вектора
    first_new = []
    for vector in product_matrix:
        part = 0
        for number in vector:
            part +=number
        first_new.append(part)
    return first_new

first_step = tran_matrix_column_vector_product(reference_matrix, example_vector)

print(first_step)
```

```python
#2. Транспонировали матрицу, полученную в предыдущем действии (производить преобразование программно не требуется)
# Значение матрицы взять из первого шага
```

```python
# Третий шаг - транспонируем эталонную матрицу и умножим ее на нетранспонированную версию

#tran_reference_matrix = matrix_transposition(reference_matrix)

#print(len(tran_reference_matrix))
#print(len(tran_reference_matrix[1]))

#умножение транспонированной матрицы на исходную подразумевает скалярное произведение каждого вектора на себя и остальные
#результаты записываются в строку
def tran_matrix_and_matrix_product(matrix):
    matrix_product_raw = []
    matrix_product_cell = []
    matrix_product = []
    count = 0
    for vector_1 in reference_matrix:
        for vector_2 in reference_matrix:
            product_cell = 0
            for a, b in zip(vector_1, vector_2):
                c = a*b
                product_cell += c
            matrix_product_raw.append(product_cell)
    for product in matrix_product_raw:
        if count < len(reference_matrix):
            matrix_product_cell.append(product)
            count+=1
        else:
            matrix_product.append(matrix_product_cell)
            matrix_product_cell = []
            matrix_product_cell.append(product)
            count=1
    matrix_product.append(matrix_product_cell)
    return matrix_product
        

third_step = tran_matrix_and_matrix_product(reference_matrix) # Полученное значение мы будем использовать в итоговом цикле
print(third_step)
print(len(reference_matrix))
```

```python
# 4. Получаем обратную матрицу

fourth_step = np.linalg.inv(third_step) # Полученное значение мы будем использовать в итоговом цикле
print(fourth_step)
```

```python
#Пятый шаг, переносим функцию из предыдущего шага и находим скалярное произведение анализируемого вектора как строки на столбец
def scalar_product_vector_column_and_row(target_vector, list_vector):
    scal_product = 0
    for num_v, num_l in zip(target_vector, list_vector):
        num_v = int(num_v)
        num_l = int(num_l)
        increment=num_v*num_l
        scal_product +=increment
    return scal_product
        

fith_step = scalar_product_vector_column_and_row(example_vector, example_vector)

print(fith_step)
```

```python
#Шестой шаг, находим итоговое значение

def final_step(first_step, fourth_step, fith_step):
    #умножаем вектор-строку на матрицу 2x2, для этого умножаем вектор на каждый столбец полученной матрицы
    sixth_one = []
    for vector in fourth_step:
        increm = 0
        for number_1, number_2 in zip(first_step, vector):
            number_3 = number_1 * number_2
            increm +=number_3
        sixth_one.append(increm)
    # Находим скалярное произведение полученного вектора-строки на вектор-столбец 
    sixth_two = 0
    for number_1, number_2 in zip(sixth_one, first_step):
        number_3 = number_1 * number_2
        sixth_two +=number_3
    #итоговое деление
    sixth_three = sixth_two/fith_step
    return sixth_three

final = final_step(first_step, fourth_step, fith_step)
print(final)
```

```python
# Теперь проанализируем все имеющиеся векора и найдем наиболее схожий
final_results = []
for vector_analise in glioma_hor_vectors:
    first_step = tran_matrix_column_vector_product(reference_matrix, vector_analise)
    fith_step = scalar_product_vector_column_and_row(vector_analise, vector_analise)
    final_new = final_step(first_step, fourth_step, fith_step)
    final_results.append(final_new)

print(final_results)

best_criteria = 9999999999999999999999999999999999999999999999
criteria_number = -1
current_best_number_0 = 0

for criteria in final_results:
    criteria_number +=1
    if criteria < best_criteria:
        best_criteria = criteria
        current_best_number_0 = criteria_number

print(current_best_number_0)

#тут мы корректируем номер вектора по имеющимся номерам
current_best_number = glioma_non_core_vector_numbers[current_best_number_0]


print(best_criteria)
print(current_best_number)
```
